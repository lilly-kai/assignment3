# 1. How much money was funded in total to the Philippines, Cambodia and Pakistan, what is the mean loan size, and how many loans is this based on?

```python
selected_country = loans[loans['country'].isin(['Cambodia', 'Philippines', 'Pakistan'])]
```

```python
selected_country['country'].value_counts()
```

```python
selected_country.groupby('country').agg(sum_loan=('funded_amount', 'sum'), mean_loan=('funded_amount', 'mean'), n_loan=('id', 'size')).sort_values('sum_loan', ascending=False)
```

The Philippines were funded the most of the three countries, about 54.5 million USD. The mean loan is highest in Cambodia (around 540 USD) and lowest in the Philippines (around 340 USD). This shows that the Philippines received many small loans (160,441 loans in total).

# 2.How much money was requested and how much was funded in total in the Education sector per year?

```python
selected_sector= loans[loans['sector'] == 'Education'].copy()
```

```python
selected_sector['year'] = selected_sector['date'].apply(lambda g: g.year)
```

```python
selected_sector[['date', 'year']].head()
```

```python
selected_sector.groupby('year').agg(funded_loan=('funded_amount', 'sum'), requested_loan=('loan_amount','sum')).sort_values('year', ascending=True)
```

You can see that the funds went up in the Education sector (except for 2017, which is not complete). The requests also went up. It's clear that the requests were a bit more than the funds in each year.

# 4.Which ten countries got the most loans for repairs?

```python
repair_loans=loans[loans['use'].str.contains('repair', case=False, na=False)]
```

```python
len(repair_loans)
```

```python
repair_loans.groupby('country').agg(n_loan=('id', 'size')).sort_values('n_loan',ascending=False).head(10)
```

The result shows that the Philippines got the most loans for repairs, but compared with question 1 it makes sense, since they got the most loans in total. If you calculate the percentage of repair loans from total loans, you can see that only 1.4% were for repairs in the Philippines, but 6.3% in Pakistan.

# 5.Do loans from a single male or female person have a higher median number of lenders?

```python
selected_loans = loans[loans['borrower_genders'].isin(['female', 'male'])]
```

```python
selected_loans['borrower_genders'].value_counts()
```

```python
selected_loans.groupby('borrower_genders').agg(lender_median=('lender_count','median')).sort_values('lender_median',ascending=False)
```

Male borrowers have a higher median number of lenders than female borrowers. Interestingly, far more loans went to female borrowers than to male borrowers.
